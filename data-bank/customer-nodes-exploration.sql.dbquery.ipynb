```sql
-- Question 1: How many unique nodes are there on the Data Bank system?
SELECT count(distinct node_id) FROM customer_nodes;

-- Question 2: What is the number of nodes per region?
SELECT r.region_name, count(distinct cn.node_id) as total_nodes FROM customer_nodes cn 
JOIN regions r ON r.region_id = cn.region_id
GROUP BY r.region_name;

-- Question 3: How many customers are allocated to each region?
SELECT r.region_name, count(distinct cn.customer_id) as allocated_customers 
FROM customer_nodes cn
JOIN regions r ON r.region_id = cn.region_id
GROUP BY r.region_name;

-- Question 4: How many days on average are customers reallocated to a different node?
select AVG(datediff(day,start_date, end_date)) as days_on_average from customer_nodes
where end_date < '9999-12-31';

-- Question 5: What is the median, 80th and 95th percentile for this same reallocation days metric for each region?

SELECT r.region_name,
percentile_cont(0.5) WITHIN GROUP (ORDER BY datediff(day,start_date, end_date)) as median,
percentile_cont(0.8) WITHIN GROUP (ORDER BY datediff(day,start_date, end_date)) as p80,
percentile_cont(0.95) WITHIN GROUP (ORDER BY datediff(day,start_date, end_date)) as p95
FROM customer_nodes cn
JOIN regions r ON r.region_id = cn.region_id
WHERE end_date < '9999-12-31'
GROUP BY r.region_name

```
