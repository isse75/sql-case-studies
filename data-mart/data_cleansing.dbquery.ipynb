```sql
-- Data Cleansing steps 
DROP TABLE IF EXISTS clean_weekly_sales;

CREATE TABLE clean_weekly_sales AS
WITH base AS (
  SELECT
    to_date(week_date, 'd/M/yy') AS week_date,
    region,
    platform,
    segment,
    transactions,
    sales
  FROM weekly_sales
)

SELECT
  week_date,
  ceil(day(week_date) / 7.0)                      AS week_number,        -- 1 to 5
  month(week_date)                                AS month_number,       -- 1 to 12
  year(week_date)                                 AS calendar_year,      -- 4 digit year

  region,
  platform,
  CASE
    WHEN segment IS NULL OR segment = 'null' THEN 'unknown'
    ELSE segment
  END                                             AS segment,
  CASE
    WHEN segment LIKE '_1' THEN 'Young Adults'
    WHEN segment LIKE '_2' THEN 'Middle Aged'
    WHEN segment LIKE '_3' OR segment LIKE '_4' THEN 'Retirees'
    ELSE 'unknown'
  END                                             AS age_band,
  CASE
    WHEN segment LIKE 'C_' THEN 'Couples'
    WHEN segment LIKE 'F_' THEN 'Families'
    ELSE 'unknown'
  END                                             AS demographic,
  transactions,
  sales,
  round(sales / nullif(transactions, 0), 2)       AS avg_transactions
FROM base;
```
